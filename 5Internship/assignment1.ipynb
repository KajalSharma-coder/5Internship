```python
#1. Python program to calculate total marks and percentage

name = input("Enter student name: ")
student_class = input("Enter class: ")

mark1 = float(input("Enter marks of Subject 1: "))
mark2 = float(input("Enter marks of Subject 2: "))
mark3 = float(input("Enter marks of Subject 3: "))
mark4 = float(input("Enter marks of Subject 4: "))
mark5 = float(input("Enter marks of Subject 5: "))

total = mark1 + mark2 + mark3 + mark4 + mark5
percentage = (total / 500) * 100

print("\nResult:")
print("Name:", name)
print("Class:", student_class)
print("Total Marks:", total)
print("Percentage:", percentage, "%")
```

```python
# 2. Python program to concatenate two strings
# perform different string methods

# Input two strings
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")
result = str1 + " " + str2
print(result)

# String methods

print("lower()      :", result.lower())
print("upper()      :", result.upper())
print("title()      :", result.title())
print("swapcase()   :", result.swapcase())
print("capitalize() :", result.capitalize())
print("casefold()   :", result.casefold())
print("center()     :", result.center(50, '*'))
print("count('a')   :", result.count('a'))
print("endswith('!'):", result.endswith('!'))
print("find('a')    :", result.find('a'))
print("isalnum()    :", result.isalnum())
print("isdigit()    :", result.isdigit())
print("isnumeric()  :", result.isnumeric())
print("isspace()    :", result.isspace())
print("replace()    :", result.replace('e', '@'))
```

```python
#3. Python program to practice Assignment Operators

x = 10
print("Initial value of x =", x)
x += 5
print("After x += 5  :", x)
x -= 3
print("After x -= 3  :", x)
x *= 2
print("After x *= 2  :", x)
x /= 4
print("After x /= 4  :", x)
x %= 3
print("After x %= 3  :", x)
x **= 2
print("After x **= 2 :", x)
x //= 2
print("After x //= 2 :", x)
```

```python
#4.  Python program to calculate total, percentage and grade

name = input("Enter student name: ")
student_class = input("Enter class: ")

mark1 = float(input("Enter marks of Subject 1: "))
mark2 = float(input("Enter marks of Subject 2: "))
mark3 = float(input("Enter marks of Subject 3: "))
mark4 = float(input("Enter marks of Subject 4: "))
mark5 = float(input("Enter marks of Subject 5: "))

total = mark1 + mark2 + mark3 + mark4 + mark5
percentage = (total / 500) * 100

if percentage >= 60:
    grade = 'A'
elif percentage >= 50 and percentage < 60:
    grade = 'B'
elif percentage >= 40 and percentage < 50:
    grade = 'C'
elif percentage >= 33 and percentage < 40:
    grade = 'D'
else:
    grade = 'Fail'

print("\nResult")
print("Name       :", name)
print("Class      :", student_class)
print("Total Marks:", total)
print("Percentage :", percentage, "%")
print("Grade      :", grade)
```
